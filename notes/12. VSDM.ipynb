# 由矩阵定义的向量空间

令 A 为 $M_{km}\mathbb{F}$ 的成员，也就是一个 k 行 n 列的矩阵。A 的行空间时 A 的行线性扩张得到的空间。它构成了n维行向量的向量子空间。A 的列空间是 A 所有列向量的线性生成子空间。它是所有 k 维向量构成的向量空间的线性子空间。 A 的零（null）空间是满足向量乘法 $A_v=0$ 的 n 维向量 v 的向量子空间。最后这个子空间的矩阵乘法也满足 $A(v_1+v_2)=Av_1+Av_2$ 和 $A(cv)=c(Av)$ 的性质。


通过矩阵乘法，我们可以将矩阵 A 定义为 $F^n$ 到 $F^k$ 的函数 $V \mapsto A_v$ 。这个函数满足以下性质：

$$
A(v_1+v_2)=Av_1+Av_2
$$

和

$$
A(cv)+=c(Av)
$$


对于这个函数，零空间是A中 $F^n$ 映射到 0 的子空间集合。因为 A 和 $F^n$的第 j 个标准基的乘积是 A 的第 j 列，A 的列空间是函数 $v \mapsto A_v$ 作为值域 $F^k$ 的像。

## 定理 2.6

设 A 在 $M_{nk}(\mathbb{F})$ 中，则

$$
dim (column space(A)) + dim(null space(A)) = \#(columns of A) = n
$$

### 证明

推论 2.5 说明零空间是 $F^n$的有限维子空间。由推论 2.3c ，零空间有一个基 $\{v_1, \cdots, v_r\}$。由推论 2.3b 我们可以连接 $v_{r+1}, \cdots, v_n$ 得到 $\{v_1, \cdots, v_n\}$ 是 $F^n$ 的一个基。如果 v 在 $F^n$中，我们可以用这个基将 v 展开为 $v = c_1v_1+\cdots+c_nv_n$，应用给定的 A

$$
Av=A(c_1v_1+\cdots+c_nv_n)=c_1Av_1+\cdots+c_rAv_r+c_{r+1}A{r+1}+\cdots+c_nAv_n=c_{r+1}Av_{r+1}+c_nAv_{n}
$$

因此向量 $Av_{r+1},\cdots,Av_n$ 张成列空间。

观察列空间的基的形式，因我们假设 $c_{r+1}Av_{r+1}+,\cdots,+c_nAv_n=0$ 。那么$A(c_{r+1}v_{r+1}+\cdots+c_nv_n)=0$，且 $c_{n+1}v_{n+1}+\cdots+c_nv_n$ 是零空间。因此 $\{v_1,\cdots,v_n\}$ 是零空间的一个基。我们对适当的标量 $a_1,\cdots,a_v$ 有

$$
c_{r+1}+v{r+1}+\cdots+c_nv_n=a_1v_1+\cdots+a_rv_r
$$

因此

$$
(-a_1)v_1+\cdots+(-a_r)v_r+c_{r+1}v_{r+1}+c_nv_n=0
$$

因为 $v_1,\cdots,Av_n$ 是线性无关的，所有的 $c_j$ 为 0，因此我们知道 $Av_{r+1},\cdots,Av_n$ 是线性无关的，因此是列空间的一组基。

最终，我们的结论是 $r+(n-r)=n$，其中 $n-r$ 可以解释为 $dim(column space(A))$ ，r 可以解释为 $dim(null space(A))$。命题得证。

## 命题 2.7

如果 A 是 $M_{kn}(F)$，则它每一个行操作都保持其行空间。

### 证明

零 A 的行为 $r_1,\cdots,r_k$，如果我们交换 2 到多个非零行，不会改变其张成的空间。当 $i \neq j $ 时，我们将行 $r_i$ 替换为 $r_i+cr_j$ 不会改变其张成的空间。因为：

$$
a_ir_i+a_jr_j=a_i(r_i+cr_j)+(a_j-a_ic)r_j
$$

说明任何原行的有限线性组合都是新行的有限线性组合。而

$$
b_i(r_i+cr_j)+b_jr_j=b_ir_i+(b_jc+b_j)r_j
$$
展示了其逆操作。

## 定理 2.8

如果 $M_{kn}(F)$ 中的 A 可以约简为梯形行形式 R，那么

$$
\begin{aligned}
dim(row\, space(A)) &= dim(row\, space(R)) \\
&= \text{\#}(nonzero\, rows\, of\, R) \\
&= \text{\#}(corner\, variables\, of\, R)\\
and \\
dim(null\, space(A)) &= dim(null\, space(R)) = \text{\#}(independent\, variables\, of\, R)
\end{aligned}
$$


### 证明

第一部分的等式可以由命题 2.7 立刻得到，最后一部分等式可以由行约简方法得到。为了得到中间的等式，我们需要确定 R 的非零行是线性无关的。令R的非零行为 $r_1,\cdots,r_t$。对于每一个 $1 \le i \le t$ 中的 j，对于 $r_i$ 的第一个非零索引，我们根据 1.5 节表示为 $j(i)$。由行约简形式的定义，此项应为1，其它行此项应为0。如果一个有限线性组合 $c_1r_1+\cdots+c_tr_t$ 为0，这就意味着 $j(i)^{th}$ 项得到 $c_i=0$，因此我们得到所有的系数为0。这就证明了我们前面说的线性无关性。
线性方程组的行操作不改变其解集，第二部分的第一个等式，可以由1.5节的结论得到。为了了解第二部分等式，我们回顾一下，解的形式是一组特定向量的有限线性组合，系数是线性无关的。第二个方程表示解空间的基向量形式。我们因此可以证明他们是线性无关的。令独立变量为 $x_j$，令相对应的向量为 $v_j$ ，那么我们知道向量 $v_j$ 的 $j^{th}$ 项为1，其它向量的 $j^{th}$ 为0。 如果这些向量的有限线性组合为 0，那么考察 $j^{th}$ 项，说明它系数为 0.结论得证。

## 推论 2.9

如果 A 属于 $M_{kn}{F}$ ，那么

$$
dim(row\,space(A))+dim(null\,space(A))=\text{\#}(columns\,of\,A)=n
$$

### 证明

我们根据定理2.8添加两个形式

$$
dim(row space(A)+ dim(null\,space(A))
$$

等于 $\text{\#}(corner variables of R) + \text{\#}(independent variables of $)$，因此所有的变量都是对角变量或独立变量，其和为 n。命题得证。

## 推论2.10

如果 A 属于 $M_{kn}F$，那么

$$
dim(row\,space(A))=dim(column space(A))
$$

### 注记

A 和行空间的维度值，和列空间的维度值，被称作 A 的秩。有些作者使用独立的术语“行秩”和“列秩”称呼两者，但是它们的值是一样的。

### 证明

以下过程根据 定理 2.6 和推论 2.9。

尽管证明方法初看起来比较抽象，但是所有的证明步骤通过计算都可以明确的推导出来，所有计算方法的根基都是行约简。定义 A 的转置为 $A'$，如 A 属于 $M_{nk}(F)$ ，那么 $A'$ 属于 $M_{kn}(F)$，且有 $(A^t)_{ij}=A{ji}$。特别是，A 的转置的行向量是 A 的列向量，反之亦然。

计算方法
1. A 的行空间的基：行约简 A，取梯形阵的非零行。
2. A 的列空间的基：转置 A，通过算法 1 计算 $A^t$ 的行空间的基，将得到的行向量转置为A的列向量。
3. A 的零空间的基：使用 1.5 节的求解过程求 $Av=0$。解集给出了列向量的所有有限线性组合，系数是独立变量。得到的列向量给出了零空间的一个基。
4. 列向量 $v_1,\cdots,v_n$的线性张成的基。将列组合为矩阵 A，那么 A 的线性张成和基可以由方法2得到。
5. 扩展 $\mathbb{F}^n$ 的线性独立集合 $v_1,\cdots,v_n$ 到 $\mathbb{F}^n$ 的基。将列排列为矩阵，转置并行约简。附加额外的行，每个自变量一个，具体如下：如果 $x_j$ 是一个自变量，那么 $x_j$ 的对应的向量的第 j 项 $j^{th}$ 系数为 1，其它项为 0。将这些向量转置为列向量，那么它们就成为一个基。
6. 选出列向量集合 ${v_1,\cdots,v_r}$ 的一个子集，成为为 ${v_1,\cdots,v_r}$ 的线性张成集的一个基。对 $0\le i\le r$ 中的每个 i，计算 $d_i=dim(span{v_1,\dots,v_r})$。如果 $d_{i-1} \le d_i$，那么保留它，否则丢弃。
